```python
import numpy as np
import random

def pad_to_length(arr, target_length):
    if len(arr) < target_length:
        # 计算需要填充的长度
        padding_length = target_length - len(arr)
        
        # 生成一个长度为 padding_length 的填充数组，每次填充时 d 都是随机选择的
        padding = np.array([])

        # 每次生成一个填充数，选择 d 的值为 2 或 4
        for _ in range(padding_length):
            d = random.choice([2, 4])  # 随机选择 d 的值
            if len(padding) == 0:
                padding = np.array([arr[-1] + d])  # 第一个填充值是从原数组最后一个元素加 d
            else:
                padding = np.append(padding, padding[-1] + d)  # 后续填充值根据上一个填充值加 d
        
        # 将原数组和填充数组连接起来
        return np.concatenate([arr, padding])
    
    # 如果数组的长度已经足够，直接返回原数组
    return arr
```

```python
# 测试数据
arr = np.array([1, 2, 3])

# 目标长度
target_length = 6

# 调用函数进行填充
result = pad_to_length(arr, target_length)

print(result)
```
